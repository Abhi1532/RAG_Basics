## **ADDING NEW DOCS TO EXISTING** ##

#### **STEP 1: Extracting text from new PDFs** ####

```python
from langchain_community.document_loaders import PyPDFLoader 
```

```python
new_pdf_path='./Docs/fabric-admin.pdf'
loader=PyPDFLoader(new_pdf_path)
docs=loader.load()
```

#### **STEP 2: Chunking the Documents** ####

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
chunks=splitter.split_documents(docs)
```

#### **STEP 3: Creating Embeddings for the CHUNKS** ####

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embedding_model=OpenAIEmbeddings(openai_api_base='https://openrouter.ai/api/v1',model="openai/text-embedding-3-small")
```

#### **STEP4: Store Embeddings of the chunks in Existing Local Vector DB** ####

```python
from langchain_community.vectorstores import Chroma
```

```python
vectorstore=Chroma(embedding_function=embedding_model,persist_directory='./VectorStore/')
vectorstore.add_documents(chunks)
```

```python
context=vectorstore.similarity_search("Why do we need fabric admin?",k=5)
```
